```python
from datasets import load_dataset
import pandas as pd

# 1. Cargar el dataset CLINC150 (usamos la versión 'plus' que tiene todos los datos)
print("Cargando el dataset...")
dataset = load_dataset("clinc_oos", "plus")

# 2. Convertir la partición de entrenamiento ('train') a un DataFrame de Pandas
# (Pandas hace que filtrar datos sea mucho más fácil)
df_train = dataset['train'].to_pandas()

# 3. Las intenciones están como números. Vamos a obtener los nombres en texto.
# La librería nos permite acceder a la lista de nombres exactos.
nombres_intenciones = dataset['train'].features['intent'].names

# Creamos una nueva columna con el nombre de la intención en texto
df_train['nombre_intencion'] = df_train['intent'].apply(lambda x: nombres_intenciones[x])

# 4. Definir la lista de intenciones del dominio de cocina que queremos mantener
intenciones_cocina = [
    'ingredient_substitution', # Sustitución de ingredientes
    'recipe',                  # Búsqueda de recetas
    'cook_time',               # Tiempo de cocción
    'nutrition_info',          # Información nutricional
    'calories',                # Calorías
    'meal_suggestion',         # Sugerencia de comida
    'ingredients_list',
    'food_last',

]

# 5. ¡Filtrar el dataset!
# Nos quedamos solo con las filas cuya intención esté en nuestra lista
dataset_cocina = df_train[df_train['nombre_intencion'].isin(intenciones_cocina)]

# Mostrar los primeros resultados para comprobar que funciona
print(f"Total de ejemplos originales: {len(df_train)}")
print(f"Total de ejemplos de cocina: {len(dataset_cocina)}")
print("\nEjemplos del dataset filtrado:")
print(dataset_cocina[['text', 'nombre_intencion']].head(10))
```
