````python
import spacy

# 1. Definimos tus MÚLTIPLES anclas (tus "radares" especializados)

nlp_en = spacy.load("en_core_web_md")
nlp_es = spacy.load("es_core_news_md")


vectores_es = {
    "comida_base": nlp_es("cocina gastronomía receta comida"),
    "origen_cultura": nlp_es("italiana mexicano japonés país cultura tradicional"),
    "tipo_dieta": nlp_es("vegano vegetariano celíaco saludable ligero"),
    "ingrediente": nlp_es("ingredientes lácteo carne pescado verdura tubérculo fruta"),
    "utensilios": nlp_es("sartén cacerola batidora cubiertos olla"),
    "electrodomestico": nlp_es("vitrocerámica nevera congelador horno freidora tostadora"),
    # "tecnicas": nlp_es("freir hornear asar saltear gratinar tostar sofreir"),
}

vectores_en = {
    "comida_base": nlp_en("cooking gastronomy recipe food"),
    "origen_cultura": nlp_en("Italian Mexican Japanese country culture traditional"),
    "tipo_dieta": nlp_en("vegan vegetarian celiac healthy light"),
    "ingrediente": nlp_en("ingredients dairy meat fish vegetable tuber fruit"),
    "utensilios": nlp_en("frying pan saucepan blender cutlery pot"),
    "electrodomestico": nlp_en("ceramic hob fridge freezer oven frier toaster"),
    # "tecnicas": nlp_en("fry bake roast saute grill toast stir-fry"),
}

````

````python

def comprobar_negacion_y_sustitucion_es(doc, palabra):
    # 1. Negaciones pegadas ("sin leche")
    if palabra.i > 0 and doc[palabra.i - 1].text.lower() in ["no", "sin", "ningún", "cero"]:
        return True

    # 2. Escalar el árbol buscando negaciones ("no quiero usar leche")
    for ancestro in palabra.ancestors:
        for hijo in ancestro.children:
            if hijo.text.lower() in ["no", "nunca", "jamás", "tampoco"]:
                return True

    # 3. NUEVO: Detección de sustitución ("sustituir leche por nata")
    # Verbos clave que indican un cambio
    verbos_cambio = ["sustituir", "cambiar", "reemplazar", "quitar"]

    # Comprobamos si algún ancestro (o la cabeza directa) es uno de estos verbos
    if palabra.head.lemma_ in verbos_cambio:
        # Si la palabra es el Objeto Directo (lo que se quita), es negativa.
        # En spaCy, la dependencia sintáctica se mira con .dep_
        if palabra.dep_ == "obj":
            return True

    return False

def comprobar_negacion_y_sustitucion_en(doc, palabra):
    if palabra.i > 0 and doc[palabra.i - 1].text.lower() in ["no", "without", "none", "zero"]:
        return True

    for ancestor in palabra.ancestors:
        for child in ancestor.children:
            if child.text.lower() in ["no", "never", "ever", "neither"]:
                return True

    change_verbs = ["replace", "change", "substitute", "remove"]

    if palabra.head.lemma_ in change_verbs:
        if palabra.dep_ == "obj":
            return True

    return False

def comprobar_negacion_y_sustitucion(doc, palabra, lang="es"):
    if lang == "es":
        return comprobar_negacion_y_sustitucion_es(doc, palabra)
    else:
        return comprobar_negacion_y_sustitucion_en(doc, palabra)


def extraer_multi_ancla(consulta_usuario, nota_corte=0.4, neg=False, nlp = nlp_es, vectores=vectores_es, lang="es"): # Por defecto asume idioma español
    doc = nlp(consulta_usuario)
    resultados = {nombre_categoria: [] for nombre_categoria in vectores.keys()}

    for palabra in doc:
        if palabra.pos_ in ["NOUN", "VERB", "ADJ", "PROPN"] and not palabra.is_stop:

            mejor_categoria = None
            mejor_puntuacion = 0

            for nombre_categoria, vector_ancla in vectores.items():
                similitud = palabra.similarity(vector_ancla)

                # Si supera el corte y es la mejor puntuación que ha sacado
                if similitud >= nota_corte and similitud > mejor_puntuacion:
                    mejor_puntuacion = similitud
                    mejor_categoria = nombre_categoria

            # Si ha encajado en alguna categoría, la guardamos ahí
            if mejor_categoria:
                es_negativa = comprobar_negacion_y_sustitucion(doc, palabra, lang=lang) and neg
                # Formateamos el resultado
                lema_final = f"SIN_{palabra.lemma_}" if es_negativa else palabra.lemma_
                resultados[mejor_categoria].append(lema_final)

    return resultados

# --- PRUEBA ---
#frase = "quiero cocinar comida italiana para sorprender a mi novia con leche keto sin horno"
#print(extraer_multi_ancla(frase))
````

````python
def extraer_entidades_spacy(texto_limpio: str, lang: str) -> dict:
    """Extrae entidades culinarias usando la libreria spacy"""
    tokens = texto_limpio.lower()
    if lang == 'en':
      entidades = extraer_multi_ancla(tokens, nlp=nlp_en, vectores=vectores_en, lang="en")
    else:
      entidades =  extraer_multi_ancla(tokens)

    items = []

    for cat, vals in entidades.items():
        if vals != []: items.extend(vals)

    return items, entidades # Devuelve una lista plana con las entidades reconocidas y la vesion por "dominio"
````

````python
import warnings

def pretty_print(texto="oregano, hacer comida italiano cenar recomendar", lang="es"):
  with warnings.catch_warnings(record=True) as w:
      # Cause all warnings to always be triggered.
      warnings.simplefilter("always")

      _, entidades = extraer_entidades_spacy(texto, lang)

      print(f'\n[{lang.upper()}] {texto}')
      if entidades:
          items = []
          for cat, vals in entidades.items():
              print(f'  {cat:17s}: {vals}')
              if vals != []: items.extend(vals)
          print(f"  Entidades: {items}")
      else:
          print('  (sin entidades detectadas)')

      # Print captured warnings
      if w:
          print("\n--- Captured Warnings ---")
          for warning_item in w:
              if issubclass(warning_item.category, UserWarning) and "[W008]" in str(warning_item.message):
                  print(f"{warning_item.message}")
              else:
                  # print(f"{warning_item.message} (Category: {warning_item.category.__name__})")
                  pass
          return False
      return True
````
