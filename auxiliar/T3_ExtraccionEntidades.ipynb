````python
class ExtractorEntidades:
  """Esta clase implementa el extractor de entidades mediante similitud del coseno"""

  VECTORES = { "es" : {
      "comida_base": "cocina gastronomía receta comida",
      "origen_cultura": "italiana mexicano japonés país cultura tradicional",
      "tipo_dieta": "vegano vegetariano celíaco saludable ligero",
      "ingrediente": "ingredientes lácteo carne pescado verdura tubérculo fruta",
      "utensilios": "sartén cacerola batidora cubiertos olla",
      "electrodomestico": "vitrocerámica nevera congelador horno freidora tostadora",
      # "tecnicas": "freir hornear asar saltear gratinar tostar sofreir",
    },
  "en" : {
      "comida_base": "cooking gastronomy recipe food",
      "origen_cultura": "Italian Mexican Japanese country culture traditional",
      "tipo_dieta": "vegan vegetarian celiac healthy light",
      "ingrediente": "ingredients dairy meat fish vegetable tuber fruit",
      "utensilios": "frying pan saucepan blender cutlery pot",
      "electrodomestico": "ceramic hob fridge freezer oven frier toaster",
      # "tecnicas": "fry bake roast saute grill toast stir-fry",
    }
  }

  NEG_AND_SUST = { "es": {
        "negaciones_pegadas" : ["no", "sin", "ningún", "cero"],
        "negaciones" : ["no", "nunca", "jamás", "tampoco"],
        "verbos_cambio" : ["sustituir", "cambiar", "reemplazar", "quitar"],
      },
    "en": {
        "negaciones_pegadas" : ["no", "without", "none", "zero"],
        "negaciones" : ["no", "never", "ever", "neither"],
        "verbos_cambio" : ["replace", "change", "substitute", "remove"],
        },
  }

  def __init__(self, langs={"en":"en_core_web_md", "es": "es_core_news_md"}):
      self.nlp = {v: spacy.load(langs[v]) for v in langs}
      self.vectores = {}
      self.neg_and_sust = {}


  def add_vectors(self, vectores=VECTORES, neg_and_sust=NEG_AND_SUST):
    self.vectores = {lang: {nombre_categoria: self.nlp[lang](vectores[lang][nombre_categoria]) for nombre_categoria in vectores[lang]} for lang in vectores}
    self.neg_and_sust = {lang: {nombre_categoria: neg_and_sust[lang][nombre_categoria] for nombre_categoria in neg_and_sust[lang]} for lang in neg_and_sust}


  def _comprobar_negacion_y_sustitucion(self, doc, palabra, lang="es"):
    #  Negaciones pegadas ("sin leche")
    if palabra.i > 0 and doc[palabra.i - 1].text.lower() in self.neg_and_sust[lang]["negaciones_pegadas"]:
        return True

    # Escalar el árbol buscando negaciones ("no quiero usar leche")
    for ancestro in palabra.ancestors:
        for hijo in ancestro.children:
            if hijo.text.lower() in self.neg_and_sust[lang]["negaciones"]:
                return True

    # Comprobamos si algún ancestro (o la cabeza directa) es uno de estos verbos
    if palabra.head.lemma_ in self.neg_and_sust[lang]["verbos_cambio"]:
        # Si la palabra es el Objeto Directo (lo que se quita), es negativa.
        # En spaCy, la dependencia sintáctica se mira con .dep_
        if palabra.dep_ == "obj":
            return True
    return False

  def extract(self, consulta_usuario, nota_corte=0.4, neg=False, lang="es", print_warn=False): # Por defecto asume idioma español
      with warnings.catch_warnings(record=True) as w:

          doc = self.nlp[lang](consulta_usuario)
          result = {nombre_categoria: [] for nombre_categoria in self.vectores[lang].keys()}

          for palabra in doc:
              if palabra.pos_ in ["NOUN", "VERB", "ADJ", "PROPN"] and not palabra.is_stop:

                  mejor_categoria = None
                  mejor_puntuacion = 0

                  for nombre_categoria, vector_ancla in self.vectores[lang].items():
                      similitud = palabra.similarity(vector_ancla)

                      # Si supera el corte y es la mejor puntuación que ha sacado
                      if similitud >= nota_corte and similitud > mejor_puntuacion:
                          mejor_puntuacion = similitud
                          mejor_categoria = nombre_categoria

                  # Si ha encajado en alguna categoría, la guardamos ahí
                  if mejor_categoria:
                      es_negativa = self._comprobar_negacion_y_sustitucion(doc, palabra, lang=lang) and neg
                      # Formateamos el resultado
                      lema_final = f"SIN_{palabra.lemma_}" if es_negativa else palabra.lemma_
                      result[mejor_categoria].append(lema_final)

          items = [v for vals in result.values() for v in vals]
          if w:
              if print_warn:
                  print("\n--- Hay errores no críticos ---")
                  for warning_item in w:
                      if issubclass(warning_item.category, UserWarning) and "[W008]" in str(warning_item.message): # El error [W008] en spaCy indica que se está calculando similitud entre palabras o textos que no tienen vectores numéricos asociados.
                          print(f"Hay palabras no reconocidas en el texto: {warning_item.message}")
      return items, result # Devolvemos el diccionario plano y por categorias


  def pretty_print(self, items, entities, texto="Desglose de entidades:", lang="es", print_all=True):
    """Atrapamos los warinings y damos formato a la salida, la opción print_warn controla si reportamos los warnings de forma eplicita,
    la opción print_all controla si imprimimos todas las categorías de las entidades (en principio nos bastal solo la lista completa pero para ilustrar)"""
    print(f'\n[{lang.upper()}] {texto}')
    if entities:
        for cat, vals in entities.items():
            if print_all: print(f'  {cat:17s}: {vals}')
        print(f"  Entidades: {items}")
    else:
        print('  (sin entidades detectadas)')
````

````python
import spacy

# Definimos tus MÚLTIPLES anclas para inglés y español, lo hacemos fuera por comodidad

nlp_en = spacy.load("en_core_web_md")
nlp_es = spacy.load("es_core_news_md")


vectores_es = {
    "comida_base": nlp_es("cocina gastronomía receta comida"),
    "origen_cultura": nlp_es("italiana mexicano japonés país cultura tradicional"),
    "tipo_dieta": nlp_es("vegano vegetariano celíaco saludable ligero"),
    "ingrediente": nlp_es("ingredientes lácteo carne pescado verdura tubérculo fruta"),
    "utensilios": nlp_es("sartén cacerola batidora cubiertos olla"),
    "electrodomestico": nlp_es("vitrocerámica nevera congelador horno freidora tostadora"),
    # "tecnicas": nlp_es("freir hornear asar saltear gratinar tostar sofreir"),
}

vectores_en = {
    "comida_base": nlp_en("cooking gastronomy recipe food"),
    "origen_cultura": nlp_en("Italian Mexican Japanese country culture traditional"),
    "tipo_dieta": nlp_en("vegan vegetarian celiac healthy light"),
    "ingrediente": nlp_en("ingredients dairy meat fish vegetable tuber fruit"),
    "utensilios": nlp_en("frying pan saucepan blender cutlery pot"),
    "electrodomestico": nlp_en("ceramic hob fridge freezer oven frier toaster"),
    # "tecnicas": nlp_en("fry bake roast saute grill toast stir-fry"),
}

````

````python

def comprobar_negacion_y_sustitucion(doc, palabra, lang="es"):
    if lang == "es":
      negaciones_pegadas = ["no", "sin", "ningún", "cero"]
      negaciones = ["no", "nunca", "jamás", "tampoco"]
      verbos_cambio = ["sustituir", "cambiar", "reemplazar", "quitar"]
    else:
      negaciones_pegadas = ["no", "without", "none", "zero"]
      negaciones = ["no", "never", "ever", "neither"]
      verbos_cambio = ["replace", "change", "substitute", "remove"]

    #  Negaciones pegadas ("sin leche")
    if palabra.i > 0 and doc[palabra.i - 1].text.lower() in negaciones_pegadas:
        return True

    # Escalar el árbol buscando negaciones ("no quiero usar leche")
    for ancestro in palabra.ancestors:
        for hijo in ancestro.children:
            if hijo.text.lower() in negaciones:
                return True

    # Comprobamos si algún ancestro (o la cabeza directa) es uno de estos verbos
    if palabra.head.lemma_ in verbos_cambio:
        # Si la palabra es el Objeto Directo (lo que se quita), es negativa.
        # En spaCy, la dependencia sintáctica se mira con .dep_
        if palabra.dep_ == "obj":
            return True

    return False

def extraer_multi_ancla(consulta_usuario, nota_corte=0.4, neg=False, nlp = nlp_es, vectores=vectores_es, lang="es"): # Por defecto asume idioma español
    doc = nlp(consulta_usuario)
    resultados = {nombre_categoria: [] for nombre_categoria in vectores.keys()}

    for palabra in doc:
        if palabra.pos_ in ["NOUN", "VERB", "ADJ", "PROPN"] and not palabra.is_stop:

            mejor_categoria = None
            mejor_puntuacion = 0

            for nombre_categoria, vector_ancla in vectores.items():
                similitud = palabra.similarity(vector_ancla)

                # Si supera el corte y es la mejor puntuación que ha sacado
                if similitud >= nota_corte and similitud > mejor_puntuacion:
                    mejor_puntuacion = similitud
                    mejor_categoria = nombre_categoria

            # Si ha encajado en alguna categoría, la guardamos ahí
            if mejor_categoria:
                es_negativa = comprobar_negacion_y_sustitucion(doc, palabra, lang=lang) and neg
                # Formateamos el resultado
                lema_final = f"SIN_{palabra.lemma_}" if es_negativa else palabra.lemma_
                resultados[mejor_categoria].append(lema_final)

    return resultados
````

````python
def extraer_entidades_spacy(texto_limpio: str, lang: str) -> dict:
    """Extrae entidades culinarias usando la libreria spacy"""
    tokens = texto_limpio.lower()

    if lang == 'en':
      entidades = extraer_multi_ancla(tokens, nlp=nlp_en, vectores=vectores_en, lang="en")
    else:
      entidades =  extraer_multi_ancla(tokens)

    items = [v for vals in entidades.values() for v in vals] # Usamos comprensión mejor

    return items, entidades # Devuelve una lista plana con las entidades reconocidas y la vesion por "dominio"
````

````python
import warnings

def pretty_print(texto="oregano, hacer comida italiano cenar recomendar", lang="es", print_warn=False, print_all=True):
  """Atrapamos los warinings y damos formato a la salida, la opción print_warn controla si reportamos los warnings de forma eplicita,
  la opción print_all controla si imprimimos todas las categorías de las entidades (en principio nos bastal solo la lista completa pero para ilustrar)"""
  with warnings.catch_warnings(record=True) as w:

      warnings.simplefilter("always")

      _, entidades = extraer_entidades_spacy(texto, lang)

      print(f'\n[{lang.upper()}] {texto}')
      if entidades:
          items = []
          for cat, vals in entidades.items():
              if print_all: print(f'  {cat:17s}: {vals}')
              items.extend(vals)
          print(f"  Entidades: {items}")
      else:
          print('  (sin entidades detectadas)')

      # Imprimimos los warnings si queremos en cualquier caso si hay warnigns devolvemos false, si no, devolvemos true
      if w:
          if print_warn:
              print("\n--- Hay errores no críticos ---")
              for warning_item in w:
                  if issubclass(warning_item.category, UserWarning) and "[W008]" in str(warning_item.message): # El error [W008] en spaCy indica que se está calculando similitud entre palabras o textos que no tienen vectores numéricos asociados.
                      print(f"Hay palabras no reconocidas en el texto: {warning_item.message}")
          return False
      return True
````
